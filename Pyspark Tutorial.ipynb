Data reading from Community Catalogue

```python
df = spark.read.format('csv').option('inferSchema', 'true').option('header', 'true').load('/Volumes/workspace/sales_data/sales_volume/BigMart Sales.csv')
df.display()
```

display Schema

```python
df.printSchema()
```

Import necessary function

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

Use of select or display data

```python
df.select('Item_Identifier','Item_Weight','Item_Type').display()
```

```python
df.select(col('Item_Identifier'),col('Item_Weight'),col('Item_Type')).display()
```

Alias

```python
df.select(col('Item_Identifier').alias('Item_Id')).display()
```

Use of filter or Where

```python
# Senario 1
df.filter(col('Item_fat_content') == 'Regular').display()
```

```python
# Senario 2
df.filter((col('Item_fat_content') == 'Regular') & (col('Item_Weight') > 10)).display()

```

```python
#senario 3
df.filter((col('Outlet_Size').isNull()) & (col('Outlet_location_Type').isin('Tier 1','Tier 2'))).display()
```

Rename a column

```python
df.withColumnRenamed('Item_Identifier','Item_Id').display()
df.withColumnRenamed('Item_Weight','Item_wt').display()
```

withColumn

```python
# 1 adding a contant value column
df=df.withColumn('xxd',lit('xxx'))
```

```python
# 2 creating new column with other column
df=df.withColumn("xxa", concat(col("Item_Type").cast(StringType()), col("Item_Weight").cast(StringType())))
df=df.withColumn("xxb", df["Outlet_Establishment_Year"]+ df["Item_Outlet_Sales"])
df.display()
```

```python
# changing the values of columns
df=df.withColumn('xxc',regexp_replace(col('Item_Type'),' ','_'))
df.display()
```

Typecasting

```python
df.select(col('xxa').cast(StringType()).alias('xxa')).display()
df.select(col('xxb').cast(IntegerType()).alias('xxb')).display()
df.withColumn('xxc',col('xxc').cast(StringType()).alias('xxc')).display()
```

Sorting

```python
#sorted the 1st col in desc and 2nd col in ascending
df=df.sort(['Item_Type','Outlet_Identifier'],ascending=[0,1])
```

LIMIT

```python
df.limit(10).display()
```

Deleting or dropping column

```python
df = df.drop('xxd','xxa','xxb','xxc')
df.display()
```

Drop Duplicates

```python
# 1 All column
df.dropDuplicates().display()
df.distinct().display()
# 2 particular column
df.dropDuplicates(['Item_Type']).display()
```

Union and unionByName

```python
# for union
data1=[('01','aman'),('02','amy'),(None,'ax')]
data2=[('04','four'),('05','five'),('06','six')]
df1 = spark.createDataFrame(data1,['id','name'])
df2 = spark.createDataFrame(data2,['id','name'])
dfx = df1.union(df2)
df1.display()
df2.display()
dfx.display()
```

```python
# for unionByname
data1=[('01','aman'),('02','amy'),(None,'ax')]
data2=[('four','44'),('five','55'),('six','66')]
df1 = spark.createDataFrame(data1,['id','name'])
df2 = spark.createDataFrame(data2,['name','id'])
dfx = df1.unionByName(df2)
df1.display()
df2.display()
dfx.display()
```

String Operation

```python
# 1. initcap
df.select(initcap(col("Item_Fat_Content")).alias("Item_Fat_Content_Initcap")).distinct().display()
# 2. lower
df.select(lower(col("Item_Fat_Content")).alias("Item_Fat_Content_lower")).distinct().display()
# 3. upper
df.select(upper(col("Item_Fat_Content")).alias("Item_Fat_Content_upper")).distinct().display()
# 4. trim
df.select(trim(col("Item_Fat_Content")).alias("Item_Fat_Content_trim")).distinct().display()
# 5. ltrim
df.select(ltrim(col("Item_Fat_Content")).alias("Item_Fat_Content_ltrim")).distinct().display()
# 6. rtrim
df.select(rtrim(col("Item_Fat_Content")).alias("Item_Fat_Content_rtrim")).distinct().display()
```

DATE FUNCTIONS

```python
# 1. current date
df = df.withColumn('today_date',current_date())
df.display()
```

```python
# 2. date add
df = df.withColumn('next_week_date',date_add('today_date',7))
df.display()
```

```python
# 3. date sub
df = df.withColumn('previous_week_date',date_sub('today_date',7))
df.display()
```

```python
# 4 Date diff
df = df.withColumn('date_diff',date_diff('today_date','previous_week_date'))
df.display()
```

```python
#5 Date Format
df = df.withColumn('date_in_format',date_format('today_date','dd-MM-yyyy'))
df.display()
```
